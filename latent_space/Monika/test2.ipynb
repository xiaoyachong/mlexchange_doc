```python
"""
Test script to load and inspect a joblib file.
"""
import numpy as np
import joblib

print(f"numpy version: {np.__version__}", flush=True)

path = "../models/vae_202602/vae_joblib_test.joblib"

print(f"Loading: {path}", flush=True)
try:
    model = joblib.load(path)
    print("Loaded successfully", flush=True)
    print(f"Model type: {type(model)}", flush=True)
    print(f"Model: {model}", flush=True)
except Exception as e:
    print(f"Failed to load: {repr(e)}", flush=True)
```
